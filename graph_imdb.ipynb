```python
%matplotlib inline
import pandas as pd
%matplotlib notebook


# Exemplo de um pequeno conjunto de dados (substitua com seu DataFrame real)
dados = [
    {'Título': 'The Shawshank Redemption', 'Ano': 1994, 'Nota IMDb': 9.3},
    {'Título': 'The Godfather', 'Ano': 1972, 'Nota IMDb': 9.2},
    {'Título': 'The Dark Knight', 'Ano': 2008, 'Nota IMDb': 9.0},
    {'Título': 'Pulp Fiction', 'Ano': 1994, 'Nota IMDb': 8.9},
    {'Título': 'The Lord of the Rings: The Return of the King', 'Ano': 2003, 'Nota IMDb': 8.9},
    {'Título': 'Forrest Gump', 'Ano': 1994, 'Nota IMDb': 8.8},
    {'Título': 'Inception', 'Ano': 2010, 'Nota IMDb': 8.8}
]

# Criar DataFrame
df = pd.DataFrame(dados)

# Criar a coluna 'Década' com base no ano de lançamento
df['Década'] = (df['Ano'] // 10) * 10

# Exibir o DataFrame com a coluna 'Década'
df
```

```python
# Contar a quantidade de filmes por década
contagem_decadas = df['Década'].value_counts().sort_index()

# Exibir a contagem de filmes por década
contagem_decadas
```

```python
# Adicionar os títulos dos filmes sobre as barras (limitando a 3 filmes por década)
for decada in df['Década'].unique():
    filmes_decada = df[df['Década'] == decada]['Título'].tolist()

    # Mostrar no máximo 3 filmes por década
    for i, filme in enumerate(filmes_decada[:3]):  # Limite para exibir até 3 filmes
        plt.text(decada, i + 0.1, filme, ha='center', fontsize=9, color='black')
```
